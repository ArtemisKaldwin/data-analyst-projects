```python
#Création d'un dataframe avec la zone, l'année et l'aide alimentaire puis groupby sur zone et année 

# Sélection des colonnes nécessaires
aide_grouped = aide[['pays_beneficiaire', 'annee', 'valeur']]

# Groupement par pays bénéficiaire et année, en sommant les valeurs
aide_grouped = aide_grouped.groupby(['pays_beneficiaire', 'annee'], as_index=False).sum()

# Renommer les colonnes pour plus de clarté
aide_grouped.rename(columns={'pays_beneficiaire': 'zone', 'valeur': 'aide_alimentaire'}, inplace=True)

# Affichage du résultat
print(aide_grouped)
```

```python
#Création d'une liste contenant les 5 pays qui ont le plus bénéficiées de l'aide alimentaire

# Calcul de la somme de l'aide alimentaire par pays
top_aide_pays = aide_grouped.groupby('zone')['aide_alimentaire'].sum().reset_index()

# Tri des pays par ordre décroissant de l'aide alimentaire reçue
top_aide_pays = top_aide_pays.sort_values(by='aide_alimentaire', ascending=False)

# Extraction des 5 pays ayant reçu le plus d'aide
top_5_pays = top_aide_pays.head(5)

# Création d'une liste des noms des pays
liste_top_5_pays = top_5_pays['zone'].tolist()

# Affichage du résultat
print(liste_top_5_pays)
```

```python
#On filtre sur le dataframe avec notre liste

# Liste des pays à filtrer
pays_a_filtrer = ['République arabe syrienne', 'Éthiopie', 'Yémen', 'Soudan du Sud', 'Soudan']

# Filtrer le DataFrame pour ne garder que les pays dans la liste
aide_filtree = aide_grouped[aide_grouped['zone'].isin(pays_a_filtrer)]

# Affichage du résultat
print(aide_filtree)
```

```python
import matplotlib.pyplot as plt

# Transformation des données pour l'affichage (pivot pour le stacked bar chart)
aide_filtree_pivot = aide_filtree.pivot(index='zone', columns='annee', values='aide_alimentaire')

# Création du graphique
colors = ['#FFC0CB', '#FFD700', '#87CEFA', '#98FB98', '#DDA0DD']  # Couleurs pastel
aide_filtree_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 8),
    color=colors
)

# Paramètres du graphique
plt.title("Aide alimentaire par pays et année (barres empilées)", fontsize=14)
plt.xlabel("Pays", fontsize=12)
plt.ylabel("Aide alimentaire (en millions)", fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.legend(title="Année", fontsize=10, title_fontsize=11)
plt.tight_layout()

# Afficher le graphique
plt.show()
```
